```python
import random

game_name = "word shuffle"
word_bank = []

word_bank = ["banana","castle","puzzle","animal","garden","carrot"]

selected_word = random.choice(word_bank)
selected_word

#defining game information
incorrect_letters = []
misplaced_letters = []
max_turns = 6
used_turns = 0
selected_word

print (f"Welcome to {game_name}!")
print (f"The word to guess has {len(selected_word)} letters.")
print (f"You have {max_turns} turns to guess the word!")

import random

# 1. Setup & initialization
word_list = ["banana", "castle", "puzzle", "animal", "garden", "carrot"]
selected_word = random.choice(word_list)

max_turns = 6
used_turns = 0  # <--- Ensures variable is initialized
misplaced_letters = []
incorrect_letters = []

print(f"Welcome! The word has {len(selected_word)} letters.\n")

# 2. Main Game Loop
while used_turns < max_turns:
    guess = input("Guess a word (or type 'stop' to end game): ").lower().strip()

    if guess == "stop":
        print("Game stopped.")
        break

    # Input validation
    if len(guess) != len(selected_word) or not guess.isalpha():
        print(f"Please enter a {len(selected_word)}-letter word consisting of letters only.")
        continue

    # Increment turns (notice it's used_turns, not user_turns)
    used_turns += 1

    # Compare guess with selected_word
    for index, letter in enumerate(guess):
        if letter == selected_word[index]:
            print(letter, end=" ")
            if letter in misplaced_letters:
                misplaced_letters.remove(letter)
        elif letter in selected_word:
            if letter not in misplaced_letters:
                misplaced_letters.append(letter)
            print("_", end=" ")
        else:
            if letter not in incorrect_letters:
                incorrect_letters.append(letter)
            print("_", end=" ")

    print()

    # Win condition
    if guess == selected_word:
        print("\nCongratulations, you guessed the word!")
        break

    # Loss condition
    if used_turns == max_turns:
        print(f"\nGame over, you lost. The word was: {selected_word}")
        break

    # Summary
    print(f"Misplaced letters: {misplaced_letters}")
    print(f"Incorrect letters: {incorrect_letters}")
    print(f"You have {max_turns - used_turns} turns left!\n")
```
